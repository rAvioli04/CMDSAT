```python
import struct
import pandas as pd

def extract_data_from_binary(file_path):
    # Open the binary file
    with open(file_path, 'rb') as file:
        data = []
        
        # Read the file in chunks of 18 bytes (7 two-byte signed integers + 4-byte unsigned integer)
        while True:
            chunk = file.read(18)
            if not chunk:
                break  # End of file

            # Unpack the 7 two-byte signed integers and the 4-byte unsigned integer timestamp
            if len(chunk) == 18:
                unpacked_data = struct.unpack('7hI', chunk)  # 7 signed 2-byte integers and 1 unsigned 4-byte integer
                data.append(unpacked_data)
        
    # Convert to a pandas DataFrame
    columns = [f'Int{i+1}' for i in range(7)] + ['Timestamp']
    df = pd.DataFrame(data, columns=columns)
    
    return df

# Example usage
file_path = 'your_binary_file.bin'  # Replace with your binary file path
df = extract_data_from_binary(file_path)
print(df)
```
